```python
# GENERATE MAPPING FILES

import pandas as pd
from pathlib import Path
import shutil

dataset = "gse_58_20260529"
# root = "/mnt/HDD3/khanh/tplam/something/all_data/raw/data_drug_response"      # TCGA
root = f"/mnt/HDD3/khanh/tplam/something/all_data/raw/{dataset}"                # else
out = f"/mnt/HDD3/khanh/tplam/something/all_data/processed/{dataset}/mappings/"


for drug in Path(root).iterdir():
    for file in drug.iterdir():
        
        outdir = Path(out) / drug.name
        outdir.mkdir(parents=True, exist_ok=True)
        
        ### TCGA dataset
        if "data_drug_response" in root:
            print(f"[INFO] Processing {dataset} - {drug.name} - {file.name}")
            if "DEmiRNA" in file.name:
                df = pd.read_csv(file)
                df_filtered = df["miRNA"]
                df_filtered.to_csv(outdir / "DEmiRNA.csv", index=False)
            elif "DEmRNA" in file.name:
                df = pd.read_csv(file)
                df_filtered = df[["mRNA", "hgnc_symbol"]]
                df_filtered.to_csv(outdir / "DEmRNA.csv", index=False)
            elif "pairs" in file.name:
                outpath = Path(file).name
                shutil.copy(file, outdir / outpath)
        
        ### CCLE dataset
        elif "ccle" in root:
            print(f"[INFO] Processing {dataset} - {drug.name} - {file.name}")
            if "miRNA_selected" in file.name:
                df = pd.read_csv(file)
                # filtered list of miRNAs are the names of column of the dataframe
                df_filtered = pd.DataFrame(df.columns[1:], columns=["miRNA"])
                df_filtered.to_csv(outdir / "DEmiRNA.csv", index=False)
                print(f"[INFO] {df.shape} -> {df_filtered.shape}")
        
        ### GSE dataset
        elif "gse_20260528" in root:
            print(f"[INFO] Processing {dataset} - {drug.name} - {file.name}")
            if "miRNA_Expression_Selected" in file.name:
                df = pd.read_csv(file)
                # filtered list of miRNAs are the names of column of the dataframe
                df_filtered = pd.DataFrame(df.columns[1:], columns=["miRNA"])
                df_filtered.to_csv(outdir / "DEmiRNA.csv", index=False)
                print(f"[INFO] {df.shape} -> {df_filtered.shape}")
        
        ### GSE 58 dataset (different dataset with GSE)
        elif "gse_58" in root:
            print(f"[INFO] Processing {dataset} - {drug.name} - {file.name}")
            if "cisplatin_allmiRNA" in file.name:
                df = pd.read_csv(file)
                # get the list of miRNAs from the first row of the dataframe
                df_filtered = pd.DataFrame(df.columns[1:], columns=["miRNA"])
                df_filtered.to_csv(outdir / "DEmiRNA.csv", index=False)
                print(f"[INFO] {df.shape} -> {df_filtered.shape}")
```
