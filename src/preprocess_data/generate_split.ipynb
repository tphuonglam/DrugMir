```python
import pandas as pd
from pathlib import Path

def generate_split_file(train_path, test_path, output_path="patient_splits.csv"):
    """
    Reads train and test CSVs, extracts patient IDs, and generates a unified split file.
    """
    print(f"\n[INFO] Generating splits from {train_path} and {test_path}...")
    # Assuming the first column is the patient ID, so we use index_col=0
    df_train = pd.read_csv(train_path, index_col=0)
    df_test = pd.read_csv(test_path, index_col=0)
    
    # Create DataFrames assigning the correct split label to the indices
    df_train_split = pd.DataFrame({
        "patient_id": df_train.index,
        "split": "train"
    })
    
    df_test_split = pd.DataFrame({
        "patient_id": df_test.index,
        "split": "test"
    })
    
    # Combine the two DataFrames
    df_combined = pd.concat([df_train_split, df_test_split], ignore_index=True)
    
    # Check for duplicates (a patient accidentally in both train and test)
    duplicates = df_combined[df_combined.duplicated(subset=['patient_id'], keep=False)]
    if not duplicates.empty:
        print(f"[WARNING] Found {len(duplicates)//2} patients in BOTH train and test sets!")
        print(duplicates.head())
    
    # Save the combined file without the pandas row numbers
    df_combined.to_csv(output_path, index=False)
    
    print(f"[SUCCESS] Split file saved to: {output_path}")
    print(f"  -> Train samples: {len(df_train_split)}")
    print(f"  -> Test samples:  {len(df_test_split)}")
    print(f"  -> Total samples: {len(df_combined)}")


def combine_data_files(train_path, test_path, output_path="combined_data.csv"):
    """
    Reads train and test CSVs and combines their full data into a single file.
    """
    print(f"\n[INFO] Combining full data from {train_path} and {test_path}...")
    
    # Read the files normally (without setting index_col so we preserve all columns exactly as they are)
    df_train = pd.read_csv(train_path)
    df_test = pd.read_csv(test_path)
    
    # Concatenate the dataframes vertically
    df_combined_data = pd.concat([df_train, df_test], ignore_index=True)
    
    # Save the merged dataset
    df_combined_data.to_csv(output_path, index=False)
    
    print(f"[SUCCESS] Combined data file saved to: {output_path}")
    print(f"  -> Total rows: {len(df_combined_data)}")
    print(f"  -> Total columns: {len(df_combined_data.columns)}")


if __name__ == "__main__":
    # Define your input file paths here
    TRAIN_FILE = "/mnt/HDD3/khanh/tplam/something/all_data/raw/tcga_20260527/cisplatin/train.csv"
    TEST_FILE = "/mnt/HDD3/khanh/tplam/something/all_data/raw/tcga_20260527/cisplatin/test.csv"
    
    # Define your output file paths here
    SPLITS_OUTPUT_FILE = "/mnt/HDD3/khanh/tplam/something/all_data/processed/tcga_40_20260528/data/cisplatin/patient_splits.csv"
    COMBINED_DATA_FILE = "/mnt/HDD3/khanh/tplam/something/all_data/processed/tcga_40_20260528/data/cisplatin/cisplatin_allmiRNA.csv"
    
    # Make sure the input files actually exist before running
    if Path(TRAIN_FILE).exists() and Path(TEST_FILE).exists():
        
        # Ensure the output directory exists
        Path(SPLITS_OUTPUT_FILE).parent.mkdir(parents=True, exist_ok=True)
        
        # Run both operations
        generate_split_file(TRAIN_FILE, TEST_FILE, SPLITS_OUTPUT_FILE)
        # combine_data_files(TRAIN_FILE, TEST_FILE, COMBINED_DATA_FILE)
        
    else:
        print("[ERROR] Could not find the input files. Please check the paths.")
```
